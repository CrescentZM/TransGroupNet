```python
from yacs.config import CfgNode as CN # 使用 CfgNode 作为配置节点的基础类
import argparse # 用于从命令行读取配置文件路径
import importlib

__all__ = ['get_cfgs'] # 定义了模块的公开接口，这里只暴露 get_cfgs 函数


# 1.配置节点（Config Node）
# YACS 使用树形结构组织配置，每个节点可以包含子节点或具体参数

# 实例化根配置节点
_C = CN()
# 方法名称
_C.method = 'ours'
_C.train = True
# 数据集
_C.dataset = 'cfo'
# 调制类型列表
_C.mod_type = ["BPSK", "QPSK", "8PSK", "PAM4", "QAM16", "QAM32", "QAM64", "QAM128", "QAM256", "GFSK", "WBFM", "AM-DSB", "AM-SSB", "OOK", "4ASK", "8ASK", "16PSK", "32PSK","8APSK","GMSK", "DQPSK","16APSK","32APSK","64APSK","128APSK"]
# 数据加载的工作进程数
_C.workers = 8
_C.seed = 1
_C.gpu = 0
_C.cpu = False


# 实例化训练参数配置节点
_C.params = CN()
_C.params.network = 'TransGroupNet' # 网络模型
_C.params.loss = "loss_CE" # 损失函数类型
_C.params.batch_size = 1024 # 批量大小
_C.params.epochs = 200 # 训练轮数
_C.params.lr = 5e-3 # 学习率
_C.params.lr_decay = 0.8 # 学习率衰减率/因子
_C.params.weight_decay = 5e-2 # 权重衰减系数
_C.params.early_stop = False # 是否启用早停
_C.params.Xmode = [{"type": "APF", "options": {"IQ_norm": False, "zero_mask": False}}]

def get_cfg_defaults():
    """
    获取默认配置的深拷贝
    """
    return _C.clone()

def get_cfgs():
    """
    获取最终配置(合并默认配置和文件配置)
    """
    # 1.加载默认配置
    cfgs = get_cfg_defaults()

    # 2.创建参数解析器
    parser = argparse.ArgumentParser(description='AMR HyperParameters')

    # 3.添加--config参数
    parser.add_argument('--config',
                        type=str,
                        default='configs/cm_resnet_cfo.yaml',
                        help='type of config file. e.g. cm_resnet_cfo (configs/cm_resnet_cfo.yaml)')
    # 4.解析命令行参数
    args = parser.parse_args()

    # 5.合并配置文件
    cfgs.merge_from_file(args.config)

    return cfgs
```
